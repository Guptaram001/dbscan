# SpeedUp

```python
# Original data
labels = ["50k", "100k", "5M", "10M"]

serial_dbscan = [0.368, 0.73, 42.702, 88.583]
uf_dbscan = [2.144, 2.409, 17.213, 41.877]

# Calculate speedup
speedup = [s/u for s, u in zip(serial_dbscan, uf_dbscan)]

# Show calculated speedup values
for l, sp in zip(labels, speedup):
    print(f"{l} Speedup: {sp:.2f}")

# Plot speedup
import matplotlib.pyplot as plt

plt.figure()
plt.plot(labels, speedup, marker="o")
plt.xlabel("Dataset Size")
plt.ylabel("Speedup")
plt.title("Speedup Calculated from Original Data")
plt.show()
```

# Densired 2 dataset  Speed-up vs number of cores/nodes

```python
import matplotlib.pyplot as plt

# -------------------------
# 2 / 4 / 6 core data — Densired 2 (datasets: 50K, 100K, 5M, 10M)
# -------------------------

cores = [2, 4, 6]
dataset_labels = ["50K", "100K", "5M", "10M"]

# Serial runtime per core: [2c, 4c, 6c] for each dataset
serial_50K = [0.369, 0.36, 0.368]
serial_100K = [0.721, 0.728, 0.73]
serial_5M = [42.943, 43.538, 42.702]
serial_10M = [87.848, 87.076, 88.583]

# UF (parallel) runtime per core: [2c, 4c, 6c] for each dataset
uf_50K = [1.384, 1.809, 2.144]
uf_100K = [1.658, 1.955, 2.409]
uf_5M = [23.188, 15.348, 17.213]
uf_10M = [57.327, 40.542, 41.877]

# -------------------------
# Plot 1: Runtime vs Cores (by dataset 50K, 100K, 5M, 10M)
# -------------------------

plt.figure(figsize=(10, 6))
plt.plot(cores, serial_50K, marker='s', linestyle='--', label="Serial (50K)")
plt.plot(cores, uf_50K, marker='o', label="UF (50K)")
plt.plot(cores, serial_100K, marker='s', linestyle='--', label="Serial (100K)")
plt.plot(cores, uf_100K, marker='o', label="UF (100K)")
plt.plot(cores, serial_5M, marker='s', linestyle='--', label="Serial (5M)")
plt.plot(cores, uf_5M, marker='o', label="UF (5M)")
plt.plot(cores, serial_10M, marker='s', linestyle='--', label="Serial (10M)")
plt.plot(cores, uf_10M, marker='o', label="UF (10M)")
plt.xlabel("Number of Cores")
plt.ylabel("Runtime (seconds)")
plt.title("Runtime vs Cores by Dataset (50K, 100K, 5M, 10M)")
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(True)
plt.xticks(cores)
plt.tight_layout()
plt.show()

# -------------------------
# Plot 2: Speedup vs Cores (by dataset 50K, 100K, 5M, 10M)
# -------------------------

speedup_50K = [s/t for s,t in zip(serial_50K, uf_50K)]
speedup_100K = [s/t for s,t in zip(serial_100K, uf_100K)]
speedup_5M = [s/t for s,t in zip(serial_5M, uf_5M)]
speedup_10M = [s/t for s,t in zip(serial_10M, uf_10M)]

plt.figure(figsize=(10, 6))
plt.plot(cores, speedup_50K, marker='o', label="50K")
plt.plot(cores, speedup_100K, marker='o', label="100K")
plt.plot(cores, speedup_5M, marker='o', label="5M")
plt.plot(cores, speedup_10M, marker='o', label="10M")
plt.plot(cores, cores, linestyle='--', color='gray', label="Ideal Speedup")
plt.xlabel("Number of Cores")
plt.ylabel("Speedup")
plt.title("Speedup vs Cores by Dataset (50K, 100K, 5M, 10M)")
plt.legend()
plt.grid(True)
plt.xticks(cores)
plt.tight_layout()
plt.show()
```
